```python
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset
from sklearn.metrics import accuracy_score, f1_score, classification_report
import xgboost as xgb
```

```python
# Load embeddings from .npy files
train_embeddings = torch.tensor(np.load(r"train_embeds_tamil.npy"), dtype=torch.float32)
test_embeddings = torch.tensor(np.load(r"test_embeds_tamil.npy"), dtype=torch.float32)
y_train_path = r"tamil_data.txt"
y_test_path = r"test_labels_tamil.txt"

# Load labels, skipping the first line (comment)
y_train = np.loadtxt(y_train_path, dtype=float)
y_test = np.loadtxt(y_test_path, dtype=float)
#y_test = np.loadtxt(y_test_path, dtype=float, skiprows=1)

y_train = torch.tensor(y_train, dtype=torch.long)
y_test = torch.tensor(y_test, dtype=torch.long)


xgb_model = xgb.XGBClassifier(
    objective='binary:logistic',
    eval_metric='aucpr',
    booster='gbtree',
    n_estimators=200,  # Number of trees
    max_depth=6,        # Depth of each tree
    learning_rate=0.1,  # Learning rate
    random_state=42,
#    tree_method= "gpu_hist",
    tree_method = "hist", 
    device = "cuda"
)

# Train the XGBoost model on BERT embeddings
xgb_model.fit(train_embeddings, y_train)

# Make predictions
y_pred = xgb_model.predict(test_embeddings)


# Evaluate the model
print("Accuracy:", accuracy_score(y_test, y_pred))
print("F1 Score:", f1_score(y_test, y_pred, average="weighted"))
print("Classification Report:\n", classification_report(y_test, y_pred))
```
